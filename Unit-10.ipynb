# Scatter plot

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([0,1,2,3,2,4,11,12,14,16,3])
y = np.array([3,8,1,10,2,3,2,4,11,12,3])

plt.scatter(x,y,marker="*")

x = np.array([0,11,12,14,4,5,6,7,8,9])
y = np.array([9,7,6,5,1,4,11,12,2,8])

plt.scatter(x,y,c='k',marker="+")

x = np.array([5,6,7,0,1,2,3,4,8,9])
y = np.array([9,7,6,5,1,0,4,3,2,8])
colors = np.array(['red','green','blue','yellow','pink','magenta','orange','purple','beige','gray'])
sizes = np.array([90,70,60,50,10,60,40,30,20,80])
plt.scatter(x,y,c=colors,s=sizes)

plt.show()
```

```python
x = np.array([5,6,7,0,3,4,8,1,2,9])
y = np.array([0,3,4,5,6,7,9,8,1,2])
colors = np.array([0,10,20,40,50,60,70,80,90,100])

plt.scatter(x,y,c=colors,cmap='viridis',alpha=0.5)
plt.colorbar()
plt.show()
```

```python
x = np.random.randint(10,50,size=20)
y = np.random.randint(10,50,size=20)
colors = np.random.randint(10,50,size=20)
sizes = 10 * np.random.randint(10,50,size=20)

plt.scatter(x,y,c=colors,s=sizes,alpha=0.5,cmap='nipy_spectral')
plt.show()
```

# Pie chart

```python
import matplotlib.pyplot as plt
import numpy as np

y = np.array([35,25,45,15])
mylables = ["Apple","Banana","Cherry","Dates"]
myexplode = [0,0,0,0.1]

plt.pie(y,startangle=90,labels=mylables,explode=myexplode,shadow=True,autopct='%1.2f%%',radius=0.5)
plt.title("Pie-Chart")
plt.legend(title="Colors")
plt.tight_layout()
plt.show()
```

```python
x = np.array([0,1,2,3])
y = np.array([3,7,4,10])
plt.subplot(2,3,1)
plt.plot(x,y)
plt.title("SALES")
plt.xlabel("x")
plt.ylabel("y")

x = np.array([5,6,7,0,1,2,3,4,8,9])
y = np.array([9,7,6,5,1,0,4,3,2,8])
plt.subplot(2,3,2)
plt.scatter(x,y)
plt.title("Scatter")
plt.xlabel("x")
plt.ylabel("y")

x = np.array(['A','B','C','D'])
y = np.array([3,8,1,10])
plt.subplot(2,3,3)
plt.bar(x,y,color='#4caf50',label="ABC",width=0.4)
plt.title("Bar")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()

x = np.array(['A','B','C','D'])
y = np.array([3,8,1,10])
plt.subplot(2,3,4)
plt.barh(x,y,color='#4caf50',label="ABC")
plt.title("Barh")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()

x = np.array([1,1,2,3,3,3,4,4,5,7,8,8,9,10,10])
plt.subplot(2,3,5)
plt.hist(x)
plt.title("hist")
plt.xlabel("x")
plt.ylabel("y")

y = np.array([35,25,45,15])
plt.subplot(2,3,6)
mylables = ["Apple","Banana","Cherry","Dates"]
myexplode = [0,0,0,0.1]
plt.pie(y,startangle=90,labels=mylables,explode=myexplode,shadow=True,autopct='%1.2f%%')
plt.title("Pie-Chart")

plt.tight_layout()
plt.show()
```

# create a line plot inwhich x points varies 3 to 21 and y point varies according to equation y=2x+8.

```python
x = range(3,22)
y = [2*i+8 for i in x]
plt.title("Line plot")
plt.xlabel("x")
plt.ylabel("y=2x+8")
plt.plot(x,y)
plt.show()
```
